Assignment 34: Prime Number Function
Write a function named is_prime().

Requirements
Accept one integer.
Return True when the number is prime.
Return False otherwise.
Use the function to print all prime numbers from 1 to 100.
Hint
A prime number has exactly two factors.

```python
def is_prime(num):
  if num < 2:
    return False

  for i in range(2, int(num ** 0.5) + 1):
    if num % i == 0:
      return False

  return True

print("Prime numbers from 1 to 100:")

for num in range(1, 101):
  if is_prime(num):
    print(num, end=" ")
```

Handle zero and negative term counts.
Do not use recursion in this assignment.

Example Input
Enter number of terms: 7

Expected Output
[0, 1, 1, 2, 3, 5, 8]

```python
def generate_fibonacci(terms):
  if terms < 0:
    return "Error: Number of terms cannot be negative."
  elif terms  == 0:
    return []
  elif terms == 1:
    return [0]

  fib =  [0, 1]

  while len(fib) < terms:
    fib.append(fib[-1] + fib[-2])
  return fib

  n = int(input("Enter number of terms: "))

  result = generate_fibonacci(n)

  print(result)
```

```python
from functools import reduce

numbers = list(map(int, input("Enter numbers sepatared by spaces: ").split()))

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

squares = list(map(lambda x: x * 2, even_numbers))

total = reduce(lambda a, b: a + b, squares, 0)

print("Even numbers:", even_numbers)
print("Squares:", squares)
print("Final total:", total)

```

Return all three inner functions.
Hint
Return multiple functions:

return deposit, withdraw, check_balance


```python
def create_bank_account(balance):

  def deposit(amount):
    nonlocal balance
    balance += amount
    return balance
    print("Deposited:", amount)

  def withdraw(amount):
    nonlocal balance
    if amount > balance:
          print("Insufficient balance!")
    else:
      balance -= amount
      print("Withdrawn:", amount)

  def check_balance():
    print("Current Balance:", balance)

  return deposit, withdraw, check_balance

deposit, withdraw, check_balance = create_bank_account(1000)

check_balance()
deposit(1000)
check_balance()
withdraw(400)
check_balance()
withdraw(200)
check_balance()
```

It should print another message after execution.
Apply it to an addition function.
Preserve the original function's return value.
Expected Pattern
Starting function...
Result: 30
Function completed.

```python
def execution_logger(func):
  def wrapper(a, b):
    print("Starting function...")
    result = func(a, b)
    print("Result:", result)
    print("Function completed,")
    return result
  return wrapper

@execution_logger
def add(a, b):
  return a + b

add(30, 64)
```
